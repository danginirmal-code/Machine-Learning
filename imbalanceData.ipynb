```python
import numpy as np
import pandas as pd
np.random.seed(123)
n_samples=1000
class_0_ratio=0.9
n_class_0=int(n_samples*class_0_ratio)
n_class_1=n_samples-n_class_0
```

```python
n_class_0,n_class_1
```

##Create my dataframe with imbalanced dataset


```python
import numpy as np
class_0=pd.DataFrame({
    'feature_1':np.random.normal(loc=0,scale=1,size=n_class_0),
    'feature_2':np.random.normal(loc=0,scale=1,size=n_class_0),
    'target':[0]*n_class_0
})
class_1=pd.DataFrame({
    'feature_1':np.random.normal(loc=2,scale=1,size=n_class_1),
    'feature_2':np.random.normal(loc=2,scale=1,size=n_class_1),
    'target':[1]*n_class_1
})
```

```python
df=pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
##Up sampling
df_minority=df[df['target']==1]
df_majority=df[df['target']==0]
```

```python
from sklearn.utils import resample
minority_resample=resample(df_minority,replace=True,n_samples=len(df_majority),random_state=42)
```

```python
minority_resample.head()
```

```python
minority_resample.shape
```

```python
df_upsampled=pd.concat([df_majority,minority_resample])
```

```python
df_upsampled['target'].value_counts()
```

```python
##Down sampling
import numpy as np
class_0=pd.DataFrame({
    'feature_1':np.random.normal(loc=0,scale=1,size=n_class_0),
    'feature_2':np.random.normal(loc=0,scale=1,size=n_class_0),
    'target':[0]*n_class_0
})
class_1=pd.DataFrame({
    'feature_1':np.random.normal(loc=2,scale=1,size=n_class_1),
    'feature_2':np.random.normal(loc=2,scale=1,size=n_class_1),
    'target':[1]*n_class_1
})
df=pd.concat([class_0,class_1]).reset_index(drop=True)
print(df['target'].value_counts())
```

```python
##Up sampling
df_minority=df[df['target']==1]
df_majority=df[df['target']==0]
```

```python
from sklearn.utils import resample
majority_downsampled=resample(df_majority,replace=False,n_samples=len(df_minority),random_state=42)
```

```python
majority_downsampled.shape
```

```python
df_downsampled=pd.concat([df_minority,majority_downsampled])
```
